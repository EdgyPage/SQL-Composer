# change_a_saved_table

Changes a Saved table in the warehouse with one of the admin module's Admin statements, each logged in admin/admin_log.md.

## Terms

- **Step**: each numbered heading below, with the code cell under it. Its first line says its kind: it changes only files, or only prints; it is a call to the warehouse, yours to run; or it is a decision, yours to make.
- **Names cell**: the code cell after the one that finds the Project. It holds this Workflow's inputs, the start materials you fill in once, each with what to write in its comment; each step's call reads them.
- **Saved table**: a table in the warehouse that only your own code writes, its files in a folder of its own.
- **Admin statement**: a Statement from the Toolbox's admin module that changes a Saved table's definition or deletes its data, such as dropping some of its Partitions; it shows what it will change before its Hive.
- **Partition**: the rows of a table that share one value of each column it is partitioned by.
- **Hive**: the SQL your warehouse runs.
- **Statement**: a query your code builds, which becomes one Hive string.

## Before step 1

Fill in the names cell below and save the notebook: each call reads it, and so do `project.check()`, which names the next step, and `project.do_next()`, the cell before the last.

## Decision rules

- If you run the steps yourself, run each step's cell in turn, in order.
- If you want them run for you, run the last cell, `project.run_workflow(...)`, in place of their cells. It runs every step left in turn, up to the first only you can do, and says where it stopped: go on from that step.
- If you want only the steps that change files run for you, run `project.do_next()`, the cell before the last. It runs them up to the first it can't.

## Self-check

Once the last step is done, each of these holds:

- Step 1's cell sets `change` with its arguments, no `...` left.
- admin/sent.csv records the Admin statement on `TABLE`, and admin/admin_log.md logs it.
- You ran the check its `Check first` line names, and any check after it.
- `project.check()` names no next step of this Workflow.

```python
import pathlib
import sys
sys.path.insert(0, str(pathlib.Path.cwd().parents[2]))  # the Project: three up

from sqlglot_composer import admin, hive_settings, project, run_execute, run_query, show_hive
from warehouse import query, execute
```

```python
TABLE = "mart_dev.runs_per_day"  # the Saved table to change, such as "mart.daily_job_runs"
```

## 1. Write the Admin statement for the change

It is a decision, yours to make.

- **Goal:** `change`, the Admin statement that makes the change you want to `TABLE`.
- **Action:** In the cell below, write the Partitions to change in place of the `...` in `where=...`, or call another Admin statement in place of `admin.drop_partitions` (`help(admin)` lists them), then run the cell.
- **Done when** the cell sets `change` with its arguments, no `...` left, and runs with no error; then step 2.

```python
from sqlglot_composer import between, equals  # where= takes a Condition
table = project.table_reference(TABLE)
# such as where=equals(table.dt, "2026-09-24"), or a range of days,
# where=between(table.dt, "2026-09-01", "2026-09-30"), dt its partition column
change = admin.drop_partitions(table, where=equals(table.dt, "2026-09-24"))  # or another: help(admin)
```

## 2. Print its Hive, and run the check its Check first line names

It changes only files, or only prints: run its cell.

- **Goal:** the change's Hive read and checked before you send it.
- **Action:** Run the cell below and read the Hive it prints, then run the check its `Check first` line names, such as a `SHOW PARTITIONS`.
- **Done when** that check shows the change reaches only what you mean it to; then step 3.

```python
show_hive(change)
```

## 3. Send it, then any check after it, such as admin.check_the_copy

It is a call to the warehouse, yours to run.

- **Goal:** the change made in the warehouse, and logged.
- **Action:** Run the cell below. It sends `change` through your execute. Then run any check after it, such as `admin.check_the_copy`.
- **Done when** admin/sent.csv records the Admin statement on `TABLE`, and admin/admin_log.md logs it; the Workflow is done: go through the Self-check at the top.

```python
run_execute(change, execute)
```

```python
project.do_next()
```

```python
# Every step left, in place of their cells, up to the first only you can do.
# Add dry_run=True to see the Hive it would send, sending nothing through your
# execute: help(project.run_workflow) says the steps it still runs:
project.run_workflow("change_a_saved_table", query=query, execute=execute, change=change)
```
