# mart_dev.team_report

A Saved table made by the batch team_report, with project.convert(...), from the statements in `conversion/statement.hql`, as its script wrote them. Its logic cell builds the same query from the Toolbox's own functions, with create(), preview() and write(). Answer its open Decisions, which `conversion/checklist.md` lists, in the decisions cell of `admin/workflows/convert/convert.ipynb`, then run the cells top to bottom. The cells comparing it with the original stay until you delete them.

```python
import pathlib
import sys
sys.path.insert(0, str(pathlib.Path.cwd().parents[1]))  # the Project: two up

from sqlglot_composer import (compare_hive, compare_with_original, export_hive, export_lineage,
                              project, run_query, show_hive)
from warehouse import query
```

```python
# logic
from sqlglot_composer import (
    Statement,
    unchecked_hive,
)
from sqlglot_composer import create_table  # for create()
from sqlglot_composer import UncheckedStatement, unchecked_hive  # for write()
from references.mart_dev__team_report import team_report  # the Saved table's own Table reference


def converted() -> Statement:
    """The query, as the Hive this Conversion was made from writes it."""
    return unchecked_hive("WITH job_runs AS (SELECT r.job_id, r.duration_mins, j.team FROM ops.job_runs AS r LEFT JOIN ops.jobs AS j ON r.job_id = j.job_id), totals AS (SELECT team, COUNT(*) AS runs, SUM(duration_mins) AS total_duration_mins FROM job_runs GROUP BY team) SELECT t.team, COALESCE(s.runs, 0) AS runs, COALESCE(s.total_duration_mins, 0) AS total_duration_mins FROM (SELECT 'data' AS team UNION ALL SELECT 'finance' UNION ALL SELECT 'web') AS t LEFT JOIN totals AS s ON t.team = s.team")


def create() -> Statement:
    """Create mart_dev.team_report from its Table reference, if it isn't there yet."""
    return create_table(team_report, may_exist=True)


def preview() -> Statement:
    """The rows write() would save, as a SELECT to run and check first."""
    return converted()


def write() -> UncheckedStatement:
    """Save the rows into mart_dev.team_report, as the original's INSERT did: kept as Hive,
    as converted() is."""
    return unchecked_hive("WITH job_runs AS (SELECT r.job_id, r.duration_mins, j.team FROM ops.job_runs AS r LEFT JOIN ops.jobs AS j ON r.job_id = j.job_id), totals AS (SELECT team, COUNT(*) AS runs, SUM(duration_mins) AS total_duration_mins FROM job_runs GROUP BY team) INSERT OVERWRITE TABLE mart_dev.team_report SELECT t.team, COALESCE(s.runs, 0) AS runs, COALESCE(s.total_duration_mins, 0) AS total_duration_mins FROM (SELECT 'data' AS team UNION ALL SELECT 'finance' UNION ALL SELECT 'web') AS t LEFT JOIN totals AS s ON t.team = s.team")
```

```python
# Try it: its rows, then, once its Table reference has its location=, the Hive of
# each step
run_query(preview(), query)
show_hive([create(), write()])
```

```python
# Lineage: a new pair in lineage/, named by the time and this table
export_lineage(write())
```

```python
# Its Hive: mart_dev__team_report.hql beside this notebook, from the logic as last saved
export_hive()
```

```python
# Its Hive against the original's, as conversion/in_dev.hql holds it
ORIGINAL = pathlib.Path("conversion/in_dev.hql").read_text(encoding="utf-8")
show_hive(preview())
print(compare_hive(preview(), ORIGINAL))
```
