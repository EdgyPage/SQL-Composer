```python
import pathlib
import sys
sys.path.insert(0, str(pathlib.Path.cwd().parents[1]))  # the Project: two up

from sqlglot_composer import (compare_hive, compare_with_original, export_hive, export_lineage,
                              project, run_query, show_hive)
from warehouse import connect, query

# Your connection to the warehouse. Each notebook runs on its own, so each connects here, at
# its top, before it calls query or execute: write yours in the line below and delete its #,
# such as connect(your_library.connect(...)). A connection kept in a file outside the Project,
# such as team_connection.py, is imported by these lines, written in that line's place:
#     sys.path.append("C:/work/connections")  # the folder team_connection.py is in
#     from team_connection import conn
#     connect(conn)
import example_database
connect(example_database)
```

```python
# logic
from sqlglot_composer import (
    AS,
    Clause,
    FROM,
    GROUP_BY,
    LEFT_JOIN,
    SELECT,
    Statement,
    WHERE,
    all_of,
    derived,
    equals,
    statement,
    sum_of,
)
from references import ops
from sqlglot_composer import create_table  # for create()
from sqlglot_composer import INSERT_OVERWRITE, Statement, statement  # for write()
from references.mart__teams_pivot import teams_pivot  # the Saved table's own Table reference

job_events_2 = ops.job_events
job_events = derived("job_events", statement(
    SELECT(
        job_events_2.job_id,
        job_events_2.event_type,
        job_events_2.dt,
        AS(job_events_2.minutes / 60, "hours"),
    ),
    # CONVERSION: reads_all_partitions=True, as the original did; check it
    FROM(job_events_2, reads_all_partitions=True),
))
e = AS(job_events, "e")
o = AS(ops.job_owners, "o")
run_hours = derived("run_hours", statement(
    SELECT(
        e.job_id,
        o.team,
        e.hours,
    ),
    FROM(e),
    # CONVERSION: reads_all_partitions=True, as the original did; check it
    LEFT_JOIN(o, ON=all_of(
        equals(e.job_id, o.job_id),
        equals(e.dt, o.dt),
    ), reads_all_partitions=True),
    WHERE(equals(e.event_type, "finish")),
))


def rows() -> list[Clause]:
    """The clauses of the query, which converted() and a write share."""
    return [
        SELECT(
            run_hours.job_id,
            # CONVERSION: adds_up=True, as the original did; check it
            AS(sum_of(run_hours.hours, adds_up=True), "sum_of_hours"),
        ),
        FROM(run_hours),
        WHERE(equals(run_hours.team, "data")),
        GROUP_BY(run_hours.job_id),
    ]


def converted() -> Statement:
    """The query, as the Hive this Conversion was made from writes it."""
    return statement(*rows())


def create() -> Statement:
    """Create mart.teams_pivot from its Table reference, if it isn't there yet."""
    return create_table(teams_pivot, may_exist=True)


def preview() -> Statement:
    """The rows write() would save, as a SELECT to run and check first."""
    return converted()


def write() -> Statement:
    """Save the rows into mart.teams_pivot, as the original's INSERT did."""
    return statement(INSERT_OVERWRITE(teams_pivot), *rows())
```

```python
# Try it: its rows, then, once its Table reference has its location=, the Hive of
# each step
run_query(preview(), query)
show_hive([create(), write()])
```

```python
# Lineage: a new pair in lineage/, named by the time and this table
export_lineage(write())
```

```python
# Its Hive: mart__teams_pivot.hql beside this notebook, from the logic as last saved
export_hive()
```
