# new_table

Makes a Table folder for each new Saved table, for you to fill in and write. A Building block, a piece several notebooks share, is made with project.new_block(...). Run each step's cell in turn, or, in their place, the last cell, project.run_workflow(...), which runs every step left in turn, up to the first only you can do, and says where it stopped. First fill in the names cell below and save the notebook: each call reads it, and so do project.check(), which names the next step, and project.do_next(), the cell before the last, which runs only the steps that change files, up to the first it can't.

```python
import pathlib
import sys
sys.path.insert(0, str(pathlib.Path.cwd().parents[2]))  # the Project: three up

from sqlglot_composer import admin, hive_settings, project, run_execute, run_query, show_hive
from warehouse import query, execute
```

```python
TABLES = []  # the tables, by their names in the warehouse, such as ["mart.daily_job_runs"]
```

## 1. Make each Table folder

It changes only files, so project.do_next() runs it for you.

```python
project.new_table(TABLES)
```

## 2. Fill in each Table reference, in references/, and save it

It is a decision, yours to make.

## 3. Write each table's logic in its Table notebook, and save it

It is a decision, yours to make.

```python
for table in TABLES:
    project.load(table).write()  # says what stops it, if anything does
```

## 4. Check each table's rows with preview(), which writes nothing

It is a call to the warehouse, yours to run.

```python
for table in TABLES:
    print(run_query(project.load(table).preview(), query))
```

## 5. Create and write each in dev

It is a call to the warehouse, yours to run.

```python
for table in TABLES:
    run_execute(project.load(table).create(), execute)
project.run_tables(tables=TABLES, execute=execute)
```
