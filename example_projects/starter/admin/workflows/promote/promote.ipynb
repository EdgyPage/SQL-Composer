# promote

Carries Saved tables from your Dev Database to your Prod Database. Run each step's cell in turn. First fill in the names cell below and save the notebook: each call reads it, and so do project.check(), which names the next step, and the last cell, project.do_next(), which runs the steps it can for you, up to the first it can't: yours, one done by hand, or one only its cell runs.

```python
import pathlib
import sys
sys.path.insert(0, str(pathlib.Path.cwd().parents[2]))  # the Project: three up

from sqlglot_composer import admin, hive_settings, project, run_execute, show_hive
from warehouse import query, execute
```

```python
TABLES = []  # the dev tables, by their names in your Dev Database, such as ["mart_dev.daily_job_runs"]
```

## 1. Promote each table: a prod copy of its Table folder and Table reference

It changes only files, so project.do_next() runs it for you.

```python
project.promote(TABLES)
```

## 2. Fill in each prod Table reference's location=

It is a decision, yours to make.

## 3. Create each table in prod

It is a call to the warehouse, yours to run.

```python
for table in project.promoted_tables(TABLES):
    run_execute(project.load(table).create(), execute)
```

## 4. Seed each prod table that reads itself, once, from its original

It is a call to the warehouse, yours to run.

```python
run_execute(hive_settings("dynamic_partitions"), execute)
for table in project.promoted_tables(TABLES, seeded=True):
    run_execute(project.load(table).seed(), execute)
```

## 5. Write each table in prod

It is a call to the warehouse, yours to run.

```python
project.run_tables(tables=project.promoted_tables(TABLES), execute=execute)
```
