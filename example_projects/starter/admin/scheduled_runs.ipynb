# Scheduled runs, by hand

What no Scheduled run does each time it runs, since it would be work for nothing each time: run these cells by hand, when you want them. Name each Scheduled run to make in `NEW_RUNS`, such as `["weekly"]`: `project.new_scheduled_run(NEW_RUNS)` makes `scheduled_runs/<name>/<name>.ipynb` from the Template, for you to fill in its `<...>`. `project.check()` reports a loop among your tables, two or more each waiting on another so that none can be written first, naming the logic cell to mend, and changes nothing. `project.export_project_lineage()` writes `lineage/<time>-project.html` and its Markdown twin: every Saved table, what reads it, and which Scheduled runs write it, a table more than one writes marked and a loop drawn red. It builds each Scheduled run as its Build cell does, with its parameters' defaults, and sends nothing. Name Scheduled runs in `RUNS` for each one's own: `project.export_run_lineage(RUNS)` writes its lineage into its folder's `lineage/`, and `project.export_run_hive(RUNS)` its built Hive, in order, into `<name>.hql` beside its notebook, to read, never to edit. Name tables in `TABLES` for `project.describe_logic(TABLES)`: each logic function of their Table notebooks, its hints and its docstring's first sentence, to paste under Copilot file 18. Each list left empty is passed over. The cell under the Project's path connects to the warehouse: a Run value that reads the warehouse needs it even to build.
