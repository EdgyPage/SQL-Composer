# The Lineage state and the Lineage difference

`lineage/lineage_state.csv`, the Lineage state, records what each column of your Project holds: its own logic and the rows that count for it, and its Effective logic, the same traced through every Saved table it reads, down to the tables the Project doesn't write. A row is added each time one changes, saying when, at which commit and from which call. Each Project action that changes a table or a Table reference records it already, and so does each Table notebook's cell 5, `export_hive()`: run the first cell below after a change made any other way, such as a Table reference edited by hand. `project.check()` says when the state is behind.

The second cell writes the Lineage difference, `lineage/<time>-difference.html` and its Markdown twin: each column with an Own change, its own logic changed, and each with an Inherited change, its own logic the same but something it reads, up its chain, changed, with that chain down to the change it comes from, and when `sent.csv` shows each changed table was next written. `SINCE` takes a time, such as `"2026-11-20"`, or a commit, such as `"1a2b3c4"`: left `None`, it compares the newest recording with the one before. `TABLES` takes a table's name, such as `"mart.team_days"`, or a list of them, to show only those and the chains that reach them. Neither cell reaches the warehouse, so this notebook needs no `connect(...)`.
